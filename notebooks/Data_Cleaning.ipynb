```python
import pandas as pd
import numpy as np
import os
```

```python
os.makedirs('../data', exist_ok=True)
```

```python
raw_csv_text = """Employee_ID,Name,Department,Salary,Joining_Date
101, Alice , HR ,60000,2021-01-15
102,Bob,IT,,2020-03-10
103, Charlie ,IT,75000,2019-07-23
104,David, Finance ,50000,
104,David, Finance ,50000,
105,Eve,hr,65000,2022-11-01"""

with open('../data/employees.csv', 'w') as f:
    f.write(raw_csv_text)
```

```python
df = pd.read_csv('../data/employees.csv')

print("--- RAW DATA ---")
print(df)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
df['Name'] = df['Name'].str.strip()
df['Department'] = df['Department'].str.strip().str.upper()
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df['Joining_Date'] = df['Joining_Date'].fillna('Unknown')
```

```python
df.to_csv('../data/cleaned_employees.csv', index=False)
print("\n--- CLEANED DATA ---")
print(df)
print("\nSuccess! Cleaned dataset saved to 'data/cleaned_employees.csv'.")
```
